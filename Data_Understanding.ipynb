```python
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
orders['order_approved_at'] = pd.to_datetime(orders['order_approved_at'])
orders['order_delivered_carrier_date'] = pd.to_datetime(orders['order_delivered_carrier_date'])
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'])
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'])
```

```python
orders.info()
```

```python
products['product_category_name'] = products['product_category_name'].fillna('Unknown')
products['product_name_lenght'] = products['product_name_lenght'].fillna(0)
products['product_description_lenght'] = products['product_description_lenght'].fillna(0)
products['product_photos_qty'] = products['product_photos_qty'].fillna(0)
```

```python
print("Null values in PRODUCTS TABLE",products.isnull().sum())
```

```python
orders['purchase_year'] = orders['order_purchase_timestamp'].dt.year
orders['purchase_month'] = orders['order_purchase_timestamp'].dt.month
```

```python
print("Total number of orders -",orders['order_id'].nunique())
print("Total number of customers -",customers['customer_id'].nunique())
print("Total number of sellers -",sellers['seller_id'].nunique())
print("Total number of products -",products['product_id'].nunique())
```

```python
products_english = products.merge(
    category_translation,
    left_on='product_category_name',
    right_on='product_category_name',
    how='left'
)

print("Product categories -",products_english['product_category_name_english'].value_counts())
print(products_english.head())
```

```python
products_english['product_category_name_english'].value_counts().head()
```

```python
orders_items = orders.merge(
    order_items,
    on='order_id',
    how='left'
)
```

```python
orders_items_products = orders_items.merge(
    products,
    on='product_id',
    how='left'
)
```

```python
master_df = orders_items_products.merge(
    customers,
    on='customer_id',
    how='left'
)
```

```python
MY_DATA = master_df.merge(
    payments,
    on='order_id',
    how='left'
)
```

```python
MY_DATA.shape
```

```python
MY_DATA.head()
```

```python
MY_DATA = MY_DATA[[
    'order_id',
    'customer_id',
    'order_status',
    'order_purchase_timestamp',
    'purchase_year',
    'purchase_month',
    'product_id',
    'seller_id',
    'price',
    'freight_value',
    'payment_type',
    'payment_value',
    'customer_city',
    'customer_state'
]]
```

```python
MY_DATA.shape
```

```python
MY_DATA.head()
```

```python
total_revenue = MY_DATA['payment_value'].sum()
print("TOTAL REVENUE -",total_revenue)

total_orders = MY_DATA['order_id'].nunique()
print("TOTAL ORDERS -",total_orders)

total_customers = MY_DATA['customer_id'].nunique()
print("TOTAL CUSTOMERS -",total_customers)
```

```python
AVG_ORDER_VALUE = total_revenue / total_orders
AVG_ORDER_VALUE
```

```python
orders_by_state = MY_DATA.groupby('customer_state')['order_id'].nunique().sort_values(ascending=False)
orders_by_state.head(10)
```

```python
revenue_by_payment = MY_DATA.groupby('payment_type')['payment_value'].sum().sort_values(ascending=False)
revenue_by_payment
```
